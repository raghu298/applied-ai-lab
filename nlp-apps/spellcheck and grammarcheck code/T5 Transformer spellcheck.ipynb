```python
from transformers import pipeline

# Load a T5 model fine-tuned for grammar and spelling
corrector = pipeline(
    "text2text-generation",
    model="vennify/t5-base-grammar-correction"
)

# The model understands context
text = "I havv goood speling and i went two the store."

corrected_text = corrector(text, max_length=128)
print(corrected_text[0]['generated_text'])

# Output: 'I have good spelling and I went to the store.'
```
