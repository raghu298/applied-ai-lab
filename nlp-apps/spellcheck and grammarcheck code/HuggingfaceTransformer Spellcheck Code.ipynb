```python
from transformers import pipeline

# Initialize the spelling correction pipeline
spell_checker = pipeline("text2text-generation", model="oliverguhr/spelling-correction-english-base")
```

```python
# Example text with typos
text = "The quick brownn fox jumps over the lazyy dog."

# Add the "fix:" prefix required by some models
corrected_text = spell_checker("fix: " + text)

# Print the corrected text
print("Original text:", text)
print("Corrected text:", corrected_text[0]['generated_text'])
```

```python
# You can also define a function for easy use
def fix_spelling(text):
    return spell_checker("fix:" + text, max_length=len(text) + 50)[0]['generated_text']

# Use the function with a new sentence
new_text = "I hav a goood idea."
print(f"Corrected: {fix_spelling(new_text)}")
```
