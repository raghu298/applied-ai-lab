```python
import pkg_resources
from symspellpy import SymSpell, Verbosity
```

```python
sym_spell = SymSpell(max_dictionary_edit_distance=2, prefix_length=7)
```

```python
dictionary_path = pkg_resources.resource_filename("symspellpy", "frequency_dictionary_en_82_765.txt")
sym_spell.load_dictionary(dictionary_path, term_index=0, count_index=1)
```

```python
sym_spell.create_dictionary_entry("SymSpell", 1)

input_term = "Interet"

suggestions = sym_spell.lookup(input_term, Verbosity.CLOSEST, max_edit_distance=2)

for suggestion in suggestions:
    print(suggestion.term, suggestion.distance, suggestion.count)
```

```python
input_term = "Sentenc eCorrectionl"

suggestions = sym_spell.lookup_compound(input_term, max_edit_distance=2)

for suggestion in suggestions:
    print(suggestion.term)
```

```python
input_term = "let'sbreakthissentenceintoindividualwords"

result = sym_spell.word_segmentation(input_term)

print(f"Segmented Sentence: {result.corrected_string}")
print(f"Sum of Edit Distances: {result.distance_sum}")
print(f"Sum of Log Probabilities: {result.log_prob_sum}")
```
