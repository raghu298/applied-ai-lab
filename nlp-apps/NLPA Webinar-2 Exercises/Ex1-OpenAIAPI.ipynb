2b. Perform Sentiment Analysis with Open AI API code

```python

def analyze_sentiment_with_score(text):

    try:
        # Create a sentiment analysis prompt and send it to the GPT-4 model
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": f"Analyze the sentiment of the following text and provide a label (positive, neutral, negative) and a score from 0 to 1: {text}"}]
        )

        # Extract the model's response
        sentiment_result = response.choices[0].message.content

        # Example format of the response to be generated by the model:
        # "Label: Positive, Score: 0.9"
        return sentiment_result

    except Exception as e:
        return f"An error occurred: {str(e)}"

if __name__ == "__main__":
    # Example text for sentiment analysis
    text = "this is not a good service"

    # Perform sentiment analysis
    result = analyze_sentiment_with_score(text)
    print("Sentiment Analysis Result:", result)
```

Additional -> Upload a "Word" document as a context. Answer questions based on the document.

```python
# Install required packages
!pip install openai --upgrade
!pip install docx2txt

import docx2txt
from google.colab import files

def extract_text_from_docx(docx_path):
    """
    Extracts text from a Word document.
    """
    return docx2txt.process(docx_path)

def ask_question_with_openai(question, context):
    """
    Ask question using OpenAI model with document context.
    """
    try:
        messages = [
            {
                "role": "system",
                "content": "You are a helpful assistant that answers questions based only on the provided document."
            },
            {
                "role": "user",
                "content": f"Context:\n{context}\n\nQuestion: {question}"
            }
        ]

        response = client.chat.completions.create(
            model="gpt-5.4-mini",
            messages=messages,
            temperature=0.1,
            max_completion_tokens=512,
            top_p=0.7
        )

        return response.choices[0].message.content

    except Exception as e:
        return f"Error: {str(e)}"

def chatbot_with_document(context):
    """
    Interactive chatbot using OpenAI.
    """
    print("Hello! I am your chatbot. Ask me anything, or type 'exit' to quit.")

    while True:
        user_input = input("You: ")
        if user_input.lower() == 'exit':
            print("Goodbye!")
            break

        answer = ask_question_with_openai(user_input, context)
        print("Bot:", answer)

if __name__ == "__main__":
    print("Please upload a Word document.")
    uploaded = files.upload()

    if uploaded:
        docx_path = list(uploaded.keys())[0]
        context = extract_text_from_docx(docx_path)
        chatbot_with_document(context)
    else:
        print("No document uploaded. Exiting.")
```

4. Whisper API
We use gTTS -> "Google Translate Text-to-Speech" library in Python

```python

# Text-to-Speech Conversion (Using gTTS)
from gtts import gTTS

# Define the text to be converted to speech
text = "The capital of India is New Delhi"

# Convert the text to speech
tts = gTTS(text=text, lang='en')

# Save the audio file
tts.save("output_audio.mp3")

print("Text-to-speech conversion completed and saved as 'output_audio.mp3'.")

# Transcribe Audio with Whisper API; "rb" stands for "Read Binary"
with open("output_audio.mp3", "rb") as audio_file:
    response = client.audio.transcriptions.create(
        model="whisper-1",
        file=audio_file
    )

# Print the transcribed text
print("Transcribed Text:", response.text)
```
