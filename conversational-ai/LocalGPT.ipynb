```python
import ollama

while True:
    question = input("Ask something (or type 'exit' to quit): ")

    if question.lower() in ['exit', 'quit']:
        print("Goodbye! 👋")
        break

    response = ollama.generate(model='gemma3:1b', prompt=question)

    print("Answer:", response['response'])
```
