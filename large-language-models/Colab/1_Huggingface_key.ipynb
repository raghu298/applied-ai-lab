This notebook demonstrates different methods for securely handling API tokens, specifically focusing on accessing the Hugging Face Hub and using models from the `transformers` library. It covers:

- **Method 1: Colab Secrets**: The recommended and most secure way to store and access tokens without exposing them in the notebook code.
- **Method 2: Direct Hardcoding**: A dangerous and insecure method shown only for quick local testing (not recommended).
- **Method 3: Manual Input (getpass)**: A semi-secure method that requires manual token entry during execution.
- **Using Environment Variable**: An alternative secure approach where the token is set as an environment variable for libraries to automatically pick up.

The notebook shows how to authenticate with Hugging Face and load models using these different methods, and attempts to demonstrate text generation using a loaded model.

Method 1: Colab Secrets - BEST
- Most secure
- Token not visible in notebook
- Easy to share notebooks safely
- Centralized token management
- No accidental exposure risk

Method 2: Direct Hardcoding - DANGEROUS
- Insecure
- Token visible in code
- Risk of accidental sharing
- Use only for quick local testing

Method 3: Manual Input (getpass) - MODERATE
- Better than hardcoding
- Requires manual entry each time
- Good for one-time use
- Still visible in execution output if printed



###METHOD 1: Using Colab Secrets (RECOMMENDED & SECURE)

```python
"""
Setup Steps:
1. Click the "key" icon in the left sidebar of Colab
2. Click "Add new secret"
3. Name: HF_TOKEN (or any name you prefer)
4. Value: Paste your Hugging Face API token
5. Toggle "Notebook access" ON
6. Get your token from: https://huggingface.co/settings/tokens
"""

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from huggingface_hub import login

# Import Colab userdata module to access secrets
from google.colab import userdata

# Retrieve the Hugging Face token from Colab secrets
HF_TOKEN = userdata.get('HF_TOKEN')

# Login to Hugging Face Hub using the secret
login(token=HF_TOKEN)

print("Authenticated using Colab Secrets")

# Load model and tokenizer
model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name,token=HF_TOKEN)

model = AutoModelForCausalLM.from_pretrained(model_name,token=HF_TOKEN,torch_dtype=torch.float16,device_map="auto")

print(f"Model loaded: {model_name}")
```

```python

# Example: Text Generation


input_text = "Machine learning is revolutionizing"
inputs = tokenizer(input_text, return_tensors="pt").to(model.device)

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_length=50,
        num_return_sequences=1,
        temperature=0.7,
        do_sample=True
    )

generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(f"\nGenerated: {generated_text}")
```

##Using Secret as Environment Variable (Alternative)

```python
import os

# Set HF token as environment variable
os.environ["HF_TOKEN"] = userdata.get('HF_TOKEN')
os.environ["HUGGING_FACE_HUB_TOKEN"] = userdata.get('HF_TOKEN')

# Now HF libraries will automatically use the environment variable
# No need to pass token parameter explicitly

print("Token set as environment variable")
```

```python
# Load model and tokenizer using environment variable, do not require explicit login to hugging facse hub
model_name = "gpt2"

tokenizer_env = AutoTokenizer.from_pretrained(model_name)

model_env = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

print(f"Model loaded using environment variable: {model_name}")
```

###METHOD 3: Using getpass for Manual Input (Semi-Secure)

```python

# METHOD 3: Using getpass for Manual Input (Semi-Secure)


from getpass import getpass

# User manually enters token (not visible while typing)
HF_TOKEN_INPUT = getpass("Enter your Hugging Face token: ")

# Login with manually entered token
login(token=HF_TOKEN_INPUT)

print("Authenticated using manual input")
```
